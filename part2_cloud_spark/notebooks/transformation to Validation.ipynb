```python
df = df.withColumn(
    "effective_tax_rate",
    when(
        col("profit_before_tax") != 0,
        round(
            (
                (col("profit_before_tax") - col("net_profit"))
                / col("profit_before_tax")
            ) * 100,
            2
        )
    )
)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lag

company_window = (
    Window
    .partitionBy("company_name")
    .orderBy("year")
)

df = df.withColumn(
    "previous_sales",
    lag("sales").over(company_window)
)

df = df.withColumn(
    "revenue_growth",
    when(
        col("previous_sales") != 0,
        round(
            (
                (col("sales") - col("previous_sales"))
                / col("previous_sales")
            ) * 100,
            2
        )
    )
)

df = df.drop("previous_sales")
```

```python
df.select(
    "company_name",
    "year",
    "sales",
    "revenue_growth"
).show(100, truncate=False)
```

```python
df = df.withColumn(
    "previous_profit",
    lag("net_profit").over(company_window)
)

df = df.withColumn(
    "profit_growth",
    when(
        col("previous_profit") != 0,
        round(
            (
                (col("net_profit") - col("previous_profit"))
                / col("previous_profit")
            ) * 100,
            2
        )
    )
)

df = df.drop("previous_profit")
```

```python
df = df.withColumn(
    "previous_eps",
    lag("eps_in_rs").over(company_window)
)

df = df.withColumn(
    "eps_growth",
    when(
        col("previous_eps") != 0,
        round(
            (
                (col("eps_in_rs") - col("previous_eps"))
                / col("previous_eps")
            ) * 100,
            2
        )
    )
)

df = df.drop("previous_eps")
```

```python
from pyspark.sql.functions import row_number

start_window = (
    Window
    .partitionBy("company_name")
    .orderBy("year")
)

end_window = (
    Window
    .partitionBy("company_name")
    .orderBy(col("year").desc())
)

cagr_base = (
    df
    .withColumn(
        "start_rank",
        row_number().over(start_window)
    )
    .withColumn(
        "end_rank",
        row_number().over(end_window)
    )
)
```

```python
cagr_base.filter(col("company_name")=="3B Blackbio DX Ltd").select("company_name","start_rank","end_rank").show()
```

```python
start_df = (
    cagr_base
    .filter(col("start_rank") == 1)
    .select(
        "company_name",
        col("year").alias("start_year"),
        col("sales").alias("beginning_sales")
    )
)
```

```python
end_df = (
    cagr_base
    .filter(col("end_rank") == 1)
    .select(
        "company_name",
        col("year").alias("end_year"),
        col("sales").alias("ending_sales")
    )
)
```

```python
cagr_df = start_df.join(
    end_df,
    on="company_name",
    how="inner"
)
```

```python
cagr_df = cagr_df.withColumn(
    "years",
    col("end_year") - col("start_year")
)
```

```python
cagr_df = cagr_df.withColumn(
    "cagr",
    when(
        (col("beginning_sales") > 0) &
        (col("ending_sales") >= 0) &
        (col("years") > 0),
        round(
            (
                (col("ending_sales") / col("beginning_sales"))
                ** (1 / col("years"))
                - 1
            ) * 100,
            4
        )
    )
)
```

```python
cagr_df = cagr_df.select(
    "company_name",
    "cagr"
)
```

```python
df = df.join(
    cagr_df,
    on="company_name",
    how="left"
)
```

```python
df.select(
    "company_name",
    "year",
    "sales",
    "cagr"
).show(100, truncate=False)
```

```python
df = df.withColumn(
    "depreciation_ratio",
    when(
        col("sales") != 0,
        round(
            (col("depreciation") / col("sales")) * 100,
            2
        )
    )
)
```

```python

company_window = (
    Window
    .partitionBy("company_name")
    .orderBy("year")
)

df = df.withColumn(
    "previous_operating_profit",
    lag("operating_profit").over(company_window)
)

df = df.withColumn(
    "operating_profit_growth",
    when(
        col("previous_operating_profit") != 0,
        (
            (col("operating_profit") - col("previous_operating_profit"))
            / col("previous_operating_profit")
        ) * 100
    )
)

df = df.withColumn(
    "operating_leverage",
    when(
        col("revenue_growth") != 0,
        round(
            col("operating_profit_growth") / col("revenue_growth"),
            2
        )
    )
)

df = df.drop(
    "previous_operating_profit",
    "operating_profit_growth"
)
```

```python
from pyspark.sql.functions import col, abs

# -----------------------------
# 1. Operating Profit
# -----------------------------
op_mismatch = df.filter(
    col("sales").isNotNull()
    & col("expenses").isNotNull()
    & col("operating_profit").isNotNull()
    & (
        abs(
            col("operating_profit")
            - (col("sales") - col("expenses"))
        ) > 1
    )
)

print("Operating Profit Mismatches      :", op_mismatch.count())


# -----------------------------
# 2. Profit Margin
# -----------------------------
profit_margin_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("profit_margin").isNotNull()
    & (
        abs(
            col("profit_margin")
            - ((col("net_profit") / col("sales")) * 100)
        ) > 0.1
    )
)

print("Profit Margin Mismatches         :", profit_margin_mismatch.count())


# -----------------------------
# 3. Operating Profit Margin
# -----------------------------
opm_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("opm").isNotNull()
    & (
        abs(
            col("opm")
            - ((col("operating_profit") / col("sales")) * 100)
        ) > 0.1
    )
)

print("Operating Margin Mismatches      :", opm_mismatch.count())


# -----------------------------
# 4. PBT Margin
# -----------------------------
pbt_margin_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("pbt_margin").isNotNull()
    & (
        abs(
            col("pbt_margin")
            - ((col("profit_before_tax") / col("sales")) * 100)
        ) > 0.1
    )
)

print("PBT Margin Mismatches            :", pbt_margin_mismatch.count())


# -----------------------------
# 5. Expense Ratio
# -----------------------------
expense_ratio_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("expense_ratio").isNotNull()
    & (
        abs(
            col("expense_ratio")
            - ((col("expenses") / col("sales")) * 100)
        ) > 0.1
    )
)

print("Expense Ratio Mismatches         :", expense_ratio_mismatch.count())


# -----------------------------
# 6. Interest Coverage
# -----------------------------
interest_coverage_mismatch = df.filter(
    col("interest").isNotNull()
    & (col("interest") != 0)
    & col("interest_coverage").isNotNull()
    & (
        abs(
            col("interest_coverage")
            - (col("operating_profit") / col("interest"))
        ) > 0.1
    )
)

print("Interest Coverage Mismatches     :", interest_coverage_mismatch.count())


# -----------------------------
# 7. Interest Burden Ratio
# -----------------------------
interest_burden_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("interest_burden_ratio").isNotNull()
    & (
        abs(
            col("interest_burden_ratio")
            - ((col("interest") / col("sales")) * 100)
        ) > 0.1
    )
)

print("Interest Burden Mismatches       :", interest_burden_mismatch.count())


# -----------------------------
# 8. Tax Burden Ratio
# -----------------------------
tax_burden_mismatch = df.filter(
    col("profit_before_tax").isNotNull()
    & (col("profit_before_tax") != 0)
    & col("tax_burden_ratio").isNotNull()
    & (
        abs(
            col("tax_burden_ratio")
            - ((col("net_profit") / col("profit_before_tax")) * 100)
        ) > 0.1
    )
)

print("Tax Burden Mismatches            :", tax_burden_mismatch.count())


# -----------------------------
# 9. Effective Tax Rate
# -----------------------------
effective_tax_mismatch = df.filter(
    col("profit_before_tax").isNotNull()
    & (col("profit_before_tax") != 0)
    & col("effective_tax_rate").isNotNull()
    & (
        abs(
            col("effective_tax_rate")
            - (((col("profit_before_tax") - col("net_profit")) / col("profit_before_tax")) * 100)
        ) > 0.1
    )
)

print("Effective Tax Rate Mismatches    :", effective_tax_mismatch.count())


# -----------------------------
# 10. Depreciation Ratio
# -----------------------------
depreciation_ratio_mismatch = df.filter(
    col("sales").isNotNull()
    & (col("sales") != 0)
    & col("depreciation_ratio").isNotNull()
    & (
        abs(
            col("depreciation_ratio")
            - ((col("depreciation") / col("sales")) * 100)
        ) > 0.1
    )
)

print("Depreciation Ratio Mismatches    :", depreciation_ratio_mismatch.count())
```
