```python
from pyspark.sql import Row

company_df = spark.createDataFrame([
    Row(company_id=1, company_name="Reliance"),
    Row(company_id=2, company_name="TCS"),
    Row(company_id=3, company_name="Infosys")
])

sales_df = spark.createDataFrame([
    Row(company_id=1, sales=876396),
    Row(company_id=2, sales=225458),
    Row(company_id=4, sales=10000)
])
```

```python



spark.range(5).show()
```

```python
sales_df.show()
```

```python
joined_df=company_df.join(sales_df,on="company_id",how="inner")
```

```python
joined_df.show()
```

```python
left_df = company_df.join(
    sales_df,
    on="company_id",
    how="left"
)

left_df.show()
```

```python
right_df = company_df.join(
    sales_df,
    on="company_id",
    how="right"
)

right_df.show()
```
